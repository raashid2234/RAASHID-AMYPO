```python

# SMART STUDENT PERFORMANCE PREDICTION SYSTEM

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score,
    accuracy_score
)


# LOAD DATASET
df = pd.read_excel('Student_Performance_Dataset.xlsx')

print("Dataset Loaded Successfully!")
print(df)

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())


# TASK 1: REGRESSION MODEL


# Select input features and target variable
X_reg = df[['Study Hours', 'Attendance (%)', 'Previous Marks']]
y_reg = df['Final Marks']

# Split training and testing data
X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X_reg, y_reg, test_size=0.2, random_state=42
)

# Train Linear Regression model
linear_model = LinearRegression()
linear_model.fit(X_train_reg, y_train_reg)

print("\nLinear Regression Model Trained Successfully!")

# Predict final marks for a new student
new_student = pd.DataFrame({
    'Study Hours': [6],
    'Attendance (%)': [85],
    'Previous Marks': [75]
})

predicted_marks = linear_model.predict(new_student)

print("\nPredicted Final Marks:", round(predicted_marks[0], 2))


# TASK 2: CLASSIFICATION MODEL


# Select input features and target variable
X_class = df[['Study Hours', 'Attendance (%)', 'Previous Marks']]
y_class = df['Result']

# Split training and testing data
X_train_class, X_test_class, y_train_class, y_test_class = train_test_split(
    X_class, y_class, test_size=0.2, random_state=42
)

# Train Logistic Regression model
log_model = LogisticRegression(max_iter=1000)
log_model.fit(X_train_class, y_train_class)

print("\nLogistic Regression Model Trained Successfully!")

# Predict result for a new student
predicted_result = log_model.predict(new_student)

print("\nPredicted Class (Pass or Fail):", predicted_result[0])


# TASK 3: COMPARE BOTH MODELS


# Regression evaluation metrics
regression_pred = linear_model.predict(X_test_reg)

mae = mean_absolute_error(y_test_reg, regression_pred)
mse = mean_squared_error(y_test_reg, regression_pred)
r2 = r2_score(y_test_reg, regression_pred)

# Classification evaluation metric
classification_pred = log_model.predict(X_test_class)

accuracy = accuracy_score(y_test_class, classification_pred)

print("\nREGRESSION EVALUATION")
print("---------------------")
print("MAE:", round(mae, 2))
print("MSE:", round(mse, 2))
print("R2 Score:", round(r2, 2))

print("\nCLASSIFICATION EVALUATION")
print("-------------------------")
print("Accuracy:", round(accuracy * 100, 2), "%")


# Model comparison table
comparison = pd.DataFrame({
    'Comparison': [
        'Algorithm Used',
        'Target Variable',
        'Type of Prediction',
        'Evaluation Metrics',
        'Real-World Application'
    ],
    'Regression': [
        'Linear Regression',
        'Final Marks',
        'Numerical Prediction',
        'MAE, MSE, R2 Score',
        'Predict Student Marks'
    ],
    'Classification': [
        'Logistic Regression',
        'Result (Pass/Fail)',
        'Categorical Prediction',
        'Accuracy',
        'Predict Student Pass/Fail'
    ]
})

print("\nFINAL MODEL COMPARISON")
print("----------------------")
print(comparison.to_string(index=False))

print("\nMACHINE LEARNING ASSIGNMENT COMPLETED!")
```
